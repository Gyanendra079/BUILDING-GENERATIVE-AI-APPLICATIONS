## Large Language Models (LLMs)

Large Language Models (LLMs) are a type of artificial intelligence program designed to understand, generate, and manipulate human language. They are typically based on transformer architecture and trained on vast amounts of text data, allowing them to perform a wide range of natural language processing (NLP) tasks such as:

*   **Text Generation:** Creating coherent and contextually relevant text, from single sentences to entire articles, stories, or poems.
*   **Summarization:** Condensing long texts into shorter, concise summaries.
*   **Translation:** Translating text from one language to another.
*   **Question Answering:** Providing answers to questions based on given context or general knowledge.
*   **Code Generation:** Writing code in various programming languages.
*   **Chatbots and Conversational AI:** Engaging in natural, human-like conversations.

LLM are opensource or paid model or hybrid ones

*   **Open-Source LLMs:** Many powerful LLMs are available as open-source projects, allowing developers to use, modify, and distribute them freely.
*   **Paid/Proprietary LLMs:** Other LLMs are developed by companies and offered as proprietary services, often through APIs, which typically require payment for usage.
*   **Hybrid Models:** Some models might have open-source base versions but offer more advanced or fine-tuned versions through paid services or commercial licenses.

## Transformer Model Architectures

Transformer models can be categorized based on their architecture:

| Architecture Type | Description | Examples |
|:------------------|:------------|:---------|
| **Encoder-only** | These models are designed for understanding input text and generating rich representations. They are good for tasks that require analyzing the input in depth but don't necessarily generate long sequences of new text. | BERT, RoBERTa, DistilBERT, ELECTRA |
| **Decoder-only** | These models are designed for generating new text sequences based on a given prompt. They excel at tasks like text generation, summarization, and creative writing. | GPT (GPT-2, GPT-3, GPT-4), Llama, PaLM, Gemini |
| **Encoder-Decoder (Seq2Seq)** | These models combine both an encoder and a decoder. The encoder processes the input sequence, and the decoder generates an output sequence based on the encoder's output. They are ideal for tasks that involve transforming an input sequence into an output sequence. | T5, BART, NLLB, original Transformer model |


BERT stands for **Bidirectional Encoder Representations from Transformers**.It is a famous machine learning model created by Google for natural language processing (NLP), which helps computers understand the context and meaning of human language

GPT stands for **Generative Pre-trained Transformer**.

## BERT vs. GPT: Use Case Comparison

BERT and GPT models are both powerful Transformer-based architectures, but they are designed and optimized for different types of tasks:

The attention mechanism in Transformer models, especially Self-Attention, helps the model weigh the importance of different parts of the input sequence when processing a particular element. The core formula for calculating attention scores is:

Attention(Q, K, V) = softmax((QKᵀ) / √dₖ)V

Let's break down each component:

Q (Query): This matrix represents the current word or token for which we are trying to compute an output. Think of it as asking "What am I looking for?"

K (Key): This matrix represents all other words/tokens in the sequence, acting as a set of labels or descriptors. Think of it as answering "What do I have available?"

V (Value): This matrix represents the actual information content of all other words/tokens in the sequence. Think of it as answering "What is the actual information I can use?"

Here's how the formula works step-by-step:

QKᵀ (Query multiplied by Transposed Key): This is the core of the attention mechanism. It calculates a similarity score between the Query (the current word) and every Key (all other words). A higher score means the words are more relevant to each other.

Scaling by √dₖ (Square Root of Key Dimension): The dot product QKᵀ can produce very large values, which could push the softmax function into regions with extremely small gradients, making training difficult. Dividing by the square root of the dimension of the key vectors (dₖ) helps to stabilize the gradients.

softmax(...): The softmax function is applied to these scaled similarity scores. This normalizes the scores so that they sum up to 1, effectively turning them into probability distributions. These are the attention weights – they indicate how much focus (attention) should be placed on each word in the sequence relative to the current Query word.

Multiplication by V (Value): Finally, these attention weights are multiplied by the Value matrix. This step aggregates the information from the Value vectors, weighted by their respective attention scores. If a word has a high attention weight, its Value vector contributes more to the final output representation of the Query word. This allows the model to selectively focus on the most relevant information.

In essence, for each word in the input sequence, the attention mechanism calculates how much it should "pay attention" to all other words (including itself) based on their relevance, and then uses these attention scores to create a weighted sum of the information (values) from those other words, forming a richer representation for the current word.

Typical Transformer architectures used for the applications listed above:

* **Sentiment classification:** Generally uses Encoder-only models (like BERT, RoBERTa). These models are excellent at understanding the context and meaning of the input text to classify its sentiment.
* **NER (Named Entity Recognition):** Also primarily uses Encoder-only models. NER involves identifying and classifying named entities in text, which requires a deep understanding of the input.
* **Text summarization:** This can use both Decoder-only (for abstractive summarization, generating new sentences) or more commonly Encoder-Decoder models (like BART, T5) which take an input text and generate a summary output.
* **Text generation:** Exclusively uses Decoder-only models (like GPT series, Llama). These models are designed to generate coherent and contextually relevant new text sequences.
Relationship extraction: Typically relies on Encoder-only models, as it involves understanding the relationships between entities mentioned within an input text.
* **Language translation:** A classic sequence-to-sequence task, which is best handled by Encoder-Decoder models (like the original Transformer model, T5, NLLB). The encoder processes the source language, and the decoder generates the target language.
In brief:

**A. Encoder-only:** Good for understanding and analyzing input (e.g., classification, NER, relationship extraction).

**B. Decoder-only**: Good for generating new text (e.g., text generation).

**C. Encoder-Decoder:** Good for transforming an input sequence into an output sequence (e.g., summarization, translation).

1. RoBERTa stands for **Robustly Optimized BERT Pretraining Approach** (or Robustly Optimized BERT Approach).

* Developer : Meta AI (formerly Facebook AI) in 2019.
* Built as an optimized and improved version of Google's BERT model.Designed for natural language processing (NLP) tasks.

2. BART - BART stands for **Bidirectional and Auto-Regressive Transformers**.

* Developer: Meta AI (Facebook AI)
* Architecture: A sequence-to-sequence model that combines a bidirectional encoder (like BERT) and an autoregressive decoder (like GPT).
* Best Used For: Text generation tasks like summarisation, translation, and abstractive question answering.

3. T5 - T5 stands for **Text-to-Text Transfer Transformer**.
* Developer: Google Research
* Core Concept: Reframes all NLP tasks into a unified "text-to-text" format. For example, both translation ("translate English to German: ...") and classification ("cola sentence: ...") use the exact same model pipeline.
* Best Used For: A wide variety of tasks including translation, summarisation, classification, and question answering.

NOTE - cola sentence: stands for the Corpus of Linguistic Acceptability

4. GPT Series - GPT stands for **Generative Pre-trained Transformer**.

* Developer: OpenAI
* Core Concept: A family of autoregressive language models that use a decoder-only Transformer architecture. They are trained to predict the next token in a sequence.

* Evolution: Includes foundational models like GPT-2 and GPT-3, instruction-tuned variants like GPT-3.5 and GPT-4, and newer reasoning models like the OpenAI o1/o3 series.

5. Llama - Llama (originally stylized as LLaMA) stands for **Large Language Model Meta AI**.

* Developer: Meta AI
* Core Concept: An open-weights family of large language models designed to deliver state-of-the-art performance while requiring significantly less computing power than closed-source alternatives.
* Evolution: Has evolved through several generations, including Llama 2, Llama 3, and Llama 4, expanding into multimodal features (vision) and massive context windows.

categorized into three types:

**Language models**: These models are designed to process and understand natural language,
allowing them to perform tasks like language translation, question answering, and text
generation.

data, allowing them to perform tasks like image classification, object detection, and scene
understanding.

language and visual data, allowing them to perform tasks like text-to-image synthesis,

Some basic points on how AI applications leverage LLMs:

1.  **LLMs (Large Language Models)**: LLMs are foundational models, often trained on vast public datasets, capable of general problem-solving, such as acting as a chatbot to answer general questions.

2.  **RAG (Retrieval Augmented Generation)**: To address questions requiring domain-specific or proprietary information, LLMs can use RAG. This involves retrieving relevant data from a knowledge base (often a vector database) and using it to augment the LLM's context, leading to more informed and accurate answers.

3.  **Tools**: When an LLM (especially within an agentic framework) needs to perform actions, access real-time information, or integrate with external systems, it uses 'tools.' These are functions or integrations (like search engines, calculators, or APIs) that extend the LLM's capabilities beyond its training data.

4.  **Agentic Capabilities / Local Interaction**: Advanced AI agents can be designed with capabilities for "computer control" or local interaction. This allows them to access, read, and process files stored on local computers or servers, enabling them to work with private, localized data for tasks like document analysis or code generation.

5.  **Fine-tuning**: Fine-tuning is a process of further training a pre-trained LLM on a smaller, specific dataset to adapt its knowledge, style, or behavior to a particular task, domain, or desired output format. This is distinct from RAG, which augments context without modifying the model's weights.

6.  **Prompt Engineering**: This is the art and science of crafting effective input prompts to guide an LLM to produce desired outputs. It optimizes the interaction with an existing LLM without altering its underlying model parameters.

## Explaining the Steps of RAG (Retrieval-Augmented Generation)

Retrieval-Augmented Generation (RAG) is an architectural pattern that combines a retriever component with a large language model (LLM) to improve the LLM's ability to generate informed and accurate responses. Here are the typical steps involved:

1.  **Prepare Input Document**: This initial step involves collecting and pre-processing the data that will serve as the knowledge base for the RAG system. This data can come from various sources like articles, documents, databases, or web pages. Pre-processing often includes cleaning, formatting, and sometimes normalizing the text to ensure consistency and quality.

2.  **Chunking**: Large documents are broken down into smaller, manageable segments called "chunks." This is crucial because LLMs have a limited context window, and providing smaller, relevant pieces of information allows the model to process and utilize the data more effectively. Chunking strategies can vary, based on sentence boundaries, paragraphs, or fixed token counts, often with some overlap to maintain context.

3.  **Create Embeddings**: Each chunk of text is then converted into a numerical vector representation called an "embedding." These embeddings capture the semantic meaning of the text, meaning that chunks with similar meanings will have similar vector representations. This is typically done using pre-trained deep learning models (e.g., Sentence-BERT, OpenAI embeddings).

4.  **Store Embeddings in Vector DB**: The generated embeddings are stored in a specialized database known as a vector database (or vector store). Vector databases are optimized for efficient similarity searches, allowing for quick retrieval of relevant chunks based on their semantic proximity to a query embedding.

5.  **Similarity Search**: When a user submits a query, that query is also converted into an embedding. The RAG system then performs a similarity search in the vector database to find the top-k (e.g., top 3, top 5) most semantically similar text chunks to the query embedding. These retrieved chunks are considered the most relevant pieces of information from the knowledge base.

6.  **RAG Pipeline**: Finally, the retrieved text chunks are combined with the user's original query to form an augmented prompt. This enriched prompt is then fed into the large language model. The LLM uses this additional context to generate a more accurate, detailed, and grounded response, reducing the likelihood of hallucinations and improving factual consistency.

## Defining the Context Window

The **context window** (also known as the context length or token window) of a Large Language Model (LLM) refers to the maximum number of tokens (words, sub-words, or characters) that the model can process and attend to at any given time, both for input and output.

Think of it as the model's short-term memory or attention span. When you provide an input prompt to an LLM, the entire prompt (including the instruction and any retrieved context in RAG) must fit within this window. Similarly, the model's generated response also consumes tokens from this same context window.

**Key aspects of the context window:**

*   **Tokens**: LLMs don't typically process raw characters but rather tokens. A token can be a whole word, part of a word, or even a single character (e.g., 'hello' might be one token, while 'unbelievable' might be 'un-believe-able' as three tokens). The exact tokenization varies by model.
*   **Input + Output**: The context window limit applies to the sum of input tokens (your prompt, system instructions, retrieved documents) and output tokens (the model's generated response). If your input is very long, the maximum length of the output you can receive will be reduced.
*   **Impact on Performance**: A larger context window generally allows the model to understand more complex queries, maintain longer conversations, and incorporate more external information (like the retrieved chunks in RAG) without losing track of the conversation or relevant details. However, larger context windows typically require more computational resources (memory and processing power) and can increase inference latency and cost.
*   **Model Variations**: Different LLMs (e.g., Claude, GPT, Gemini, Llama, Mistral) have varying context window sizes, ranging from a few thousand tokens to hundreds of thousands of tokens, constantly evolving with research and development.

### 1. Simple text Storage  

For simple text storage, character-based chunking is often used. This method splits text into chunks of a fixed number of characters. For example, a document might be broken into 500-character segments. This is straightforward to implement but can cut through sentences or words, potentially losing semantic context.

### 2. Fixed-size chunking with overlap

This is a common approach where you split text into chunks of a fixed size, but each chunk also includes a certain amount of text from the previous chunk (the overlap). This helps maintain context across chunk boundaries, which is crucial for tasks like question answering or summarization where information might span across two chunks.

### 3. Sentence-based chunking

This method involves splitting text into individual sentences. It's often preferred when semantic coherence within each chunk is important. Tools like NLTK or SpaCy can be used to accurately detect sentence boundaries. This works well for tasks where responses need to be highly accurate and contextually relevant at a sentence level, such as conversational AI or precise information retrieval.

### 4. Recursive chunking

Recursive chunking attempts to split documents hierarchically, starting with larger delimiters (like document sections or paragraphs) and then falling back to smaller ones (like sentences or words) if chunks are still too large. This method is effective for structured documents or when preserving the document's logical flow is important. It helps ensure that related content stays together, providing better context for retrieval-augmented generation (RAG) models.

### 5. Content-aware / Semantic chunking

This advanced method uses NLP techniques to understand the semantic content of the text and chunk it based on topics, themes, or logical units. It might involve techniques like topic modeling, semantic similarity, or even using LLMs to identify natural breakpoints in the text. This is highly effective for complex documents where understanding the core ideas within each chunk is paramount, leading to more relevant results for advanced LLM applications.

## Advantages and disadvantages of Conversation Memory types

### ConversationBufferMemory
*   **Advantages**: Simple to implement, retains full context of the conversation, useful for short conversations where full recall is important.
*   **Disadvantages**: Can quickly consume a large number of tokens, leading to increased costs and potential to hit context window limits for long conversations.

### ConversationBufferWindowMemory
*   **Advantages**: Limits token usage by only storing the `k` most recent interactions, providing a balance between context retention and token efficiency, suitable for moderate-length conversations.
*   **Disadvantages**: Loses context of earlier parts of the conversation if they fall outside the `k` window, which might be critical for some applications.

### ConversationSummaryMemory
*   **Advantages**: Highly token-efficient for very long conversations as it only stores a summary, provides a high-level understanding of the conversation history, less likely to hit context window limits.
*   **Disadvantages**: May lose fine-grained details or specific nuances from earlier parts of the conversation as they are condensed into a summary.

# Definitions of different types of chain

### Single Chain
*   **Single Chain** - A fundamental chain that consists of a single Large Language Model (LLM) call. It's used for straightforward tasks where the LLM can provide a direct response without needing to interact with other components or tools. These are often used for simple text generation, summarization, or classification tasks.

### Simple Sequential Chain
*   **Simple Sequential Chain** - Executes a series of LLM calls in a predetermined order, where the output of one LLM call serves as the input to the next. This is useful for multi-step tasks that can be broken down into a linear sequence of operations, like summarizing text and then generating questions based on the summary.

### Sequential Chain
*   **Sequential Chain** - Similar to Simple Sequential Chain but offers more flexibility. It allows for multiple inputs and outputs between steps, not just a single string output. This makes it suitable for more complex workflows where different parts of the chain might need to pass structured data or multiple pieces of information to the next step.

### Math Chain
*   **Math Chain** - Designed to perform mathematical calculations. It typically takes a natural language math problem as input, converts it into a solvable expression, and then uses a tool (like a calculator or Python interpreter) to solve it. This is useful for scenarios where an LLM needs to perform precise arithmetic or algebraic operations.

### RAG Chain
*   **RAG (Retrieval Augmented Generation) Chain** - Combines information retrieval with text generation. When asked a question, it first retrieves relevant documents or information from a knowledge base (e.g., a vector database) and then uses an LLM to generate an answer based on both the retrieved information and its own knowledge. This helps in grounding LLM responses in factual data and reducing hallucinations.

### Router Chain
*   **Router Chain** - Determines which of several sub-chains or tools to use based on the input query. It acts as a traffic controller, directing the user's request to the most appropriate component. This is useful for building applications that can handle a variety of different user intents, each requiring a different processing path.

### LLM Router Chain
*   **LLM Router Chain** - A specific type of Router Chain where an LLM is used to decide which sub-chain or tool to route the input to. The LLM analyzes the input query and, based on its understanding, selects the best path. This provides a more intelligent routing mechanism compared to rule-based routers.

### SQL Chain
*   **SQL Chain** - Designed to interact with SQL databases. It can take a natural language query, convert it into an SQL query, execute it against a database, and then process the results. This enables LLMs to answer questions about data stored in relational databases without requiring the user to write SQL.

## Integrating HuggingFace API with CrewAI Agents

To integrate HuggingFace models with your CrewAI agents, you'll typically use the `HuggingFaceHub` LLM wrapper from Langchain. This allows your agents to leverage the vast array of models hosted on HuggingFace for their language processing and reasoning.

### Steps for Integration:
1.  **Get a HuggingFace API Token:** Sign up on [HuggingFace](https://huggingface.co/settings/tokens) and create a new access token (it's often called `HF_TOKEN`).
2.  **Securely Store the Token:** In Google Colab, it's best practice to store your API token in the Secrets Manager. Click the '🔑' icon in the left panel, add a new secret, and name it `HF_TOKEN` (or whatever you prefer).
3.  **Configure the LLM:** Use the `HuggingFaceHub` class, passing your token and specifying the desired HuggingFace model.
4.  **Assign to Agent:** Assign this configured LLM to your `crewai.Agent` instance.

Here's an example:

```python
from crewai import Agent, Task, Crew, Process
from langchain_community.llms import HuggingFaceHub
from google.colab import userdata

# 1. Get HuggingFace API Token securely
HF_TOKEN = userdata.get('HF_TOKEN') # Assuming you stored it as 'HF_TOKEN' in Colab Secrets

# 2. Configure the HuggingFace LLM
# Choose a suitable model from HuggingFace Hub (e.g., a text-generation model)
# You might need to adjust the model and parameters based on your needs
hf_llm = HuggingFaceHub(
    repo_id="google/flan-t5-large", # Example model, choose one that fits your task
    huggingfacehub_api_token=HF_TOKEN,
    task="text-generation",
    model_kwargs={
        "temperature": 0.7,
        "max_new_tokens": 256
    }
)

# 3. Define an Agent using the HuggingFace LLM
researcher = Agent(
    role='HuggingFace Powered Researcher',
    goal='Find compelling information on a given topic using HuggingFace model',
    backstory='An expert researcher capable of synthesizing information from various sources powered by advanced HuggingFace models.',
    verbose=True,
    allow_delegation=False,
    llm=hf_llm # Assign the HuggingFace LLM here
)

# Example Task for the agent
task = Task(
    description="Summarize the latest advancements in AI ethics in 200 words.",
    expected_output="A concise summary of recent developments in AI ethics.",
    agent=researcher
)

# Initialize the Crew
crew = Crew(
    agents=[researcher],
    tasks=[task],
    verbose=2,
    process=Process.sequential
)

# Kickoff the Crew (uncomment to run)
# result = crew.kickoff()
# print(result)
```
